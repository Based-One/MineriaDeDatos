# 🕸️ Sesión 1: El ADN de las Redes
Representación de Grafos: Matriz vs Lista de Adyacencia.

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
vuelos_raw = pd.read_csv("./dataassets/vuelos_10k.csv")
# Tirando los nan y definiendo el dataset limpio
vuelos_clean = vuelos_raw.dropna().copy()
display(vuelos_raw.head(3))
# mostranod que no hay nan
print(vuelos_clean.isna().sum())
# Pasando las columnas a upper case
vuelos_clean["Origen"] = vuelos_clean["Origen"].str.upper()
vuelos_clean["Destino"] = vuelos_clean["Destino"].str.upper()
# Tirando los registros que tienen self-loop
vuelos_clean = vuelos_clean[vuelos_clean["Origen"] != vuelos_clean["Destino"]]
# Tirando los registros que la distancia es negativa
vuelos_clean = vuelos_clean[vuelos_clean["Distancia_km"] > 0]
# Tirando duplicados
vuelos_clean = vuelos_clean.drop_duplicates(subset=["Origen","Destino"]) # rutas duplicadas

# Mostrando los edges y nodos
vuelos_G = nx.from_pandas_edgelist(vuelos_clean, "Origen", "Destino")
print("Number of edges",vuelos_G.number_of_edges())
print("Number of nodes", vuelos_G.number_of_nodes())
# 1. Crear un Grafo Dirigido (Ej. Instagram)
# G = nx.DiGraph()
# G.add_edges_from([('Kamila', 'Diego L.'), ('Diego L.', 'Kamila'), ('Raul', 'Ximena'), ('Ximena', 'Raul'), ("Andres","Raul"),("Andres","Ximena"),("Octavio","Diego L."),("Octavio","Raul")])

# 2. Dibujar el Grafo
# plt.figure(figsize=(100,400))
# nx.draw(vuelos_G, with_labels=True, node_color='#F39C12', node_size=2500, font_weight='bold', font_color='white', arrowsize=20, )
# plt.title('Nuestra Primera Red')
# plt.show()
```

### ⚠️ El colapso en la RAM: La Matriz de Adyacencia
Fácil de leer para las matemáticas, terrible para la memoria de la computadora en Big Data.

```python
matriz = nx.adjacency_matrix(vuelos_G).todense()
print("Nodos en orden:", vuelos_G.nodes())
print("\nMatriz de Adyacencia:")
print(matriz)
matriz_bytes = matriz.nbytes
print("Megabytes usados en la matriz adyacente: ", matriz_bytes /1e6)
dataset_total_bytes = vuelos_clean.memory_usage().sum()
print("Megabytes que usa el dataset: ", dataset_total_bytes/1e6)
```

### 🚀 La Salvación: Lista de Aristas (Edge List)
Solo guardamos las conexiones que SÍ existen.

```python
print("Lista de Aristas (Edge List):")
for edge in vuelos_G.edges():
    print(f"{edge[0]} ---> {edge[1]}")
```
