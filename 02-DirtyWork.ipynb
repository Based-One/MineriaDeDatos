```python
def desviacionEstandar(population: list) -> float:
    if not population:
        return 0.0
        
    n = len(population)
    
    mean = sum(population) / n
    
    variance = sum((x - mean) ** 2 for x in population) / n
    
    std_dev = variance ** 0.5
    
    return std_dev

```

```python
import random

random_list = [random.random() for _ in range(500_000)]
print("Random list length:", len(random_list))
```

```python
import numpy as np



print(np.std(random_list))
```
