```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv")
```

## 1. Asimetria (Skewness) en Age

```python
asimetria_age = df["Age"].skew()
print(asimetria_age)
```

El resultado de `.skew()` en Age dio positivo (como 0.39). Eso quiere decir que la distribucion esta sesgada hacia la derecha, o sea que la mayoria de los pasajeros eran jovenes/adultos jovenes y habia menos pasajeros de edad avanzada, pero esos pocos ancianos que si iban en el barco estiran la cola de la distribucion hacia la derecha. En resumen: habia mas jovenes que ancianos, y los ancianos son los que forman esa "cola" larga hacia valores altos de edad.

## 2. Curtosis (Kurtosis) en Fare

```python
curtosis_fare = df["Fare"].kurt()
print(curtosis_fare)
```

La curtosis de Fare dio un numero muy alto, como 33.4. Un valor de curtosis tan grande (muy por encima de 0, que seria una distribucion normal) significa que las colas de la distribucion son muy "gordas" o pesadas, es decir que hay valores extremos bastante alejados del centro de los datos. Esto si confirma los outliers que encontramos en clase con el metodo del IQR: los boletos de mas de 500 libras son justamente esos valores extremos que hacen que la curtosis se dispare tanto.

## 3. Tabla cruzada (Cross-Tabulation) entre Pclass y Survived

```python
tabla_cruzada = pd.crosstab(df["Pclass"], df["Survived"])
print(tabla_cruzada)
```

En la tabla cruzada se ve clarisimo un patron: en primera clase sobrevivieron mas pasajeros (136) de los que murieron (80), en segunda clase quedo bastante parejo (87 sobrevivieron vs 97 murieron), y en tercera clase la mayoria murio por mucho (372 murieron contra solo 119 que sobrevivieron). O sea que entre mas alta la clase del pasajero, mas probabilidades tenia de sobrevivir, y los de tercera clase fueron los mas afectados con diferencia.
