```python
import pandas as pd
import numpy as np

 
# Reemplaza con la URL de tu repositorio
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)
print("read csv")
print(df['Installs'])
```

```python
print("Registros antes del drop",df['App'].count())
print("Mostrando duplicados",df.duplicated(subset=["App"]).sum())
print("Dropping duplicates")
df.drop_duplicates(subset=['App'],keep='first', inplace=True)
print("después de hacer el drop",df['App'].count())
```

_Se purgaron 1181 registros_

```python
df['Installs'] = pd.to_numeric(
    df['Installs'].str.replace(',', '', regex=False).str.replace('+','', regex=False),
    errors='coerce'
)

# Aísle el registro corrupto en lugar de descartarlo a ciegas
corruptos = df[df['Installs'].isna()]
print(f"Registros descartados: {len(corruptos)}")
df = df.dropna(subset=['Installs'])
df["Installs"].fillna(0)
print(df["Installs"].mean())
```

_La media es ≈ 7 777 506.7322 después del drop_duplicates_
Aqui tuve que descartar un registro corrupto que no sé el por qué ocurrió, pero ocurrió cuando hice drop_duplicates y use errors="coerce" para que el codigo no se pare cuando haya un registro con datos invalidos, en este caso le pone NaN porque aparecia un registro con "Free" lo cual no tenía sentido 



```python
df['Price'] = df['Price'].str.replace('$','', regex=False).astype(float)
    


print(df["Price"].mean())

df["Price"].max()

print(df[df["Price"] > 200])

df= df[df['Price'] < 50 ]
print(df)
print("Aqui se ve que el max es de 47 dolares: ", df["Price"].max())

df.to_csv('playstore_limpio.csv', index=False)
```
